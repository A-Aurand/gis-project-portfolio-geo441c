### **Reflection on my first experience with ArcGIS Pro**

- Overall, I spent one hour to finish creating my first map
- Here is a table showing things that I feel excited and challening in my first experince working in ArcGIS Pro


| Things I'm excited about      | Things challenging for me  |
| ----------------------------- | -------------------------- |
| I've heard ArcGIS has many more options and tools     | Tips & tricks I learned for QGIS don't always work  |
| Being able to be flexible by knowing how to use different programs     | It feels like there's so many new tools    |
| Having useful layers already built in through The Living Atlas      | I'm not sure how to properly size maps and build the layout around them   |

- Here is how I rate my first experience working in ArcGIS Pro:
  - Data import and management
    - Adding data and folders is slightly more complex than I'm used to. I prefer to store the data outside of the project file so that way I can easily reuse it if necessary, so with ArcGIS I have to make a folder connection to import those layers. However, I feel that once the layers and data are in the project, they're pretty easy to access and maintain.
  - Map creation and annotation
    - The most difficult thing for me is managing the size of the elements as I move them around, especially text. The text is dynamic, so trying to change the size of the frame often results in the text size being bigger, and the font size control being redundant. However, with the legend, it's the opposite problem, as I struggle to get the text to a size large enough to see, while making headers and lines different sizes.
  - Navigate the software using its GUI
    - The top ribbon bar for navigation is relatively simple, and most of the tools I need are in intuitive places, and the left panel is similarly easy to use. However, the right panel is constantly changing depending on what you have highlighted, and I find myself going back and forth, trying to find the right element to highlight to access the tool I need.
